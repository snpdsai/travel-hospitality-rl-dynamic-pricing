# Dynamic Pricing using Reinforcement Learning

## Week 1: Designing the MDP and Gymnasium Environment

### Objective

The objective of this notebook is to formulate the dynamic pricing problem as a Markov Decision Process (MDP) and implement a custom Gymnasium environment representing the airline or hotel booking process.

### Tasks

- Define the pricing problem as an MDP.
- Design the state space.
- Define the action space.
- Implement the reward function.
- Develop a stochastic demand simulator.
- Build and validate the custom Gymnasium environment.
