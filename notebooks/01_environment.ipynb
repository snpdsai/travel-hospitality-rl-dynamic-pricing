## Markov Decision Process (MDP)

The dynamic pricing problem can be formulated as a Markov Decision Process (MDP), where the agent learns an optimal pricing strategy through repeated interaction with the environment.

### State

The state is represented as:

(State = [Remaining Inventory, Days Until Departure])

Example:

- Inventory = 50 seats
- Days Remaining = 20

State = (50, 20)

---

### Actions

The action is the price level selected by the agent.

For this project, we use five discrete pricing levels:

| Action | Price Multiplier |
|---------|------------------|
| 0 | 0.8 × Base Price |
| 1 | 0.9 × Base Price |
| 2 | 1.0 × Base Price |
| 3 | 1.1 × Base Price |
| 4 | 1.2 × Base Price |

---

### Reward

The reward is the revenue earned during each booking day.

Reward = Tickets Sold × Selected Price

If no customer purchases a ticket, the reward is zero.

---

### Episode

One episode represents one complete booking season, starting from the first booking day until the departure date or until all inventory has been sold.
